```python
import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
```

```python
# Fetch Data
gold_data = yf.Ticker("GLD").history(period="1mo")
silver_data = yf.Ticker("SLV").history(period="1mo")

# Normalize the prices for bettter visual 
# (Divides every price by the very first price in the series)
gold_normalized = gold_data['Close'] / gold_data['Close'].iloc[0]
silver_normalized = silver_data['Close'] / silver_data['Close'].iloc[0]

# Plot them cleanly using the Pandas index automatically
plt.figure(figsize=(10, 5)) # Makes the chart wider and easier to see
plt.plot(gold_data.index, gold_normalized, label="GLD (Normalized)", color="gold")
plt.plot(silver_data.index, silver_normalized, label="SLV (Normalized)", color="silver")

plt.title("Gold vs Silver - Past Month (Normalized)")
plt.xlabel("Date")
plt.ylabel("Relative Performance")
plt.legend() 
plt.grid(True) 
plt.show()
```
